```python
import pyautogui
import collections
import time

skip_ad_wait_time = 0.5  # seconds

CURSOR_LOC_FINDER = 1
TEST = (CURSOR_LOC_FINDER == 0)

for i in range(20):
    time.sleep(skip_ad_wait_time)

    if CURSOR_LOC_FINDER == 1:
        skip_button_location = pyautogui.position()
        print(f"The current cursor position is: {skip_button_location}")

    if TEST:
        Point = collections.namedtuple("Point", "x y")
        skip_button_location = Point(x=483, y=507)
        print(f"Skipping ad at: {skip_button_location}")
        time.sleep(skip_ad_wait_time)
        pyautogui.click(skip_button_location)
```
